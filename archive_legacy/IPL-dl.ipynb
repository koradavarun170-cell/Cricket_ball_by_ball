```
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
NUMERIC_COLS = ["ball", "batter_runs", "batter_balls", "team_runs", "over"]
CATEGORICAL_COLS = ["bat_pos", "bowler"]
preprocessor=ColumnTransformer([
    ('num',StandardScaler(),NUMERIC_COLS),
    ('cat',OneHotEncoder(handle_unknown='ignore'),CATEGORICAL_COLS),])
```

```
import torch
import torch.nn as nn
```

```
class model(nn.Module):
    def __init__(self):
        super().__init__()
        self.layers=nn.Sequential(
            nn.Linear(14,128),
            nn.ReLU(),
            nn.Linear(128,128),
            nn.ReLU(),
            nn.Linear(128,128),
            nn.ReLU(),
            nn.Linear(128,128),
            nn.ReLU(),
            nn.Linear(128,1),
            nn.Sigmoid()
        )
    def forward(self,x):
        return self.layers(x)
```

```
mymodel=model()
mymodel.train()
```

```
loss_fn=nn.BCELoss()
optimizer = torch.optim.Adam(mymodel.parameters(), lr=0.001)
```

```
from torch.utils.data import Dataset,DataLoader

class dataset(Dataset):
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __len__(self):
        return len(self.x)
    def __getitem__(self,idx):
        return self.x[idx],self.y[idx]
```

```
train_dataset1=dataset(X_train1,y_train1.reset_index(drop=True))
test_dataset1=dataset(X_test1,y_test1.reset_index(drop=True))
```

```
train_dataset2=dataset(X_train2,y_train2.reset_index(drop=True))
test_dataset2=dataset(X_test2,y_test2.reset_index(drop=True))
```

```
train_loader1=DataLoader(train_dataset1,batch_size=32,shuffle=True)
test_loader1=DataLoader(test_dataset1,batch_size=32,shuffle=True)
```

```
train_loader2=DataLoader(train_dataset2,batch_size=32,shuffle=True)
test_loader2=DataLoader(test_dataset2,batch_size=32,shuffle=True)
```

```
for a,b in train_loader1:
    print(a.shape,b.shape)
    print(a[0][0])
    break
```

```
mymodel=model()
loss_fn=nn.BCELoss()
optimizer = torch.optim.Adam(mymodel.parameters(), lr=0.001)
num_epochs = 40
for epoch in range(num_epochs):
    total_loss = 0
    correct = 0
    total = 0
    mymodel.train()
    for x, y in train_loader1:
        x = x.float()
        y = y.float().view(-1, 1)
        optimizer.zero_grad()
        y_pred = mymodel(x)
        loss = loss_fn(y_pred, y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
        predicted = (y_pred >= 0.5).float()
        correct += (predicted == y).sum().item()
        total += y.shape[0]

    accuracy = 100 * correct / total

    print(f"Epoch [{epoch+1}/{num_epochs}] "
          f"Loss: {total_loss/len(train_loader1):.4f} "
          f"Accuracy: {accuracy:.2f}%")
```

```
mymodel.eval()
correct = 0
total = 0
with torch.no_grad():
    for x, y in test_loader1:
        x = x.float()
        y = y.float().view(-1, 1)
        y_pred = mymodel(x)
        predicted = (y_pred >= 0.5).float()
        correct += (predicted == y).sum().item()
        total += y.size(0)
print(f"Test Accuracy: {100 * correct / total:.2f}%")
```

```
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```
mymodel.eval()

y_true = []
y_pred_list = []

with torch.no_grad():
    for x, y in test_loader1:
        x = x.float()

        outputs = mymodel(x)
        preds = (outputs >= 0.3).int().view(-1)

        y_true.extend(y.numpy())
        y_pred_list.extend(preds.numpy())

print("Accuracy :", accuracy_score(y_true, y_pred_list))
print("Precision:", precision_score(y_true, y_pred_list))
print("Recall   :", recall_score(y_true, y_pred_list))
print("F1 Score :", f1_score(y_true, y_pred_list))
print("Confusion Matrix:\n", confusion_matrix(y_true, y_pred_list))
```
